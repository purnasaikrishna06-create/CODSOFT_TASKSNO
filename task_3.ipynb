```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df =pd.read_csv("/content/(task2-outcome)Employee_Performance_cleaned.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
sns.set_style("whitegrid")
```

```python
plt.figure(figsize=(8,5))
department_count=df["Department"].value_counts()
plt.bar(department_count.index,department_count.values)
plt.title("Number of Employees by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.show()
```

```python
plt.figure(figsize=(10,5))
average_salary=df.groupby("Experience")["Salary"].mean()
plt.plot(average_salary.index,average_salary.values,marker="o",linewidth=3,label="Average Salary")
plt.title("Average Salary by Years of Experience")
plt.xlabel("Experience (Years)")
plt.ylabel("Average Salary")
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(7, 7))
status_count = df["Status"].value_counts()
plt.pie(status_count.values,labels=status_count.index,autopct="%1.1f%%",startangle=45)
plt.title("Employee Status Distribution")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df["Salary"],bins=10,edgecolor="black")
plt.title("Distribution of Employee Salaries")
plt.xlabel("Salary")
plt.ylabel("Number of Employees")
plt.show()
```

```python
plt.figure(figsize=(9, 5))
sns.scatterplot(data=df,x="Experience",y="Performance Score",hue="Department",palette="Set1",s=70)
plt.title("Experience vs Performance Score", fontsize=14)
plt.xlabel("Experience (Years)", fontsize=12)
plt.ylabel("Performance Score", fontsize=12)
plt.legend(title="Department")
plt.grid(True)
plt.show()
```

##Conclusion:
The visualizations provide a clear understanding of the Employee Performance dataset.

- The bar chart shows the number of employees in each department and helps compare department-wise employee distribution.
- The line chart shows how the average salary changes with years of experience.
- The pie chart represents the proportion of employees based on their employment status.
- The histogram shows the distribution of employee salaries and helps identify the common salary ranges.
- The scatter plot shows the relationship between employee experience and performance score.

Overall, these visualizations make the trends, distributions, and relationships in the employee data easy to understand and help in making meaningful observations from the dataset.
